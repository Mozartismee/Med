```python
import pandas as pd

# 假設 df_origin 和 df_sampling 已經被定義並載入

# 1. 先檢查 df_origin 的 'AID' 和 'cancer' 是否為字串，並轉換為布林值
df_origin['AID'] = df_origin['AID'].astype(str).map({'True': True, 'False': False})
df_origin['cancer'] = df_origin['cancer'].astype(str).map({'True': True, 'False': False})

# 2. 確保 NaN 值轉為 False
df_origin['AID'] = df_origin['AID'].fillna(False)
df_origin['cancer'] = df_origin['cancer'].fillna(False)

# 3. 確保 '病歷號' 兩邊格式一致（避免數據類型不匹配導致合併失敗）
df_origin['病歷號'] = df_origin['病歷號'].astype(str).str.strip()
df_sampling['病歷號'] = df_sampling['病歷號'].astype(str).str.strip()

# 4. 進行合併
df_merged = pd.merge(df_origin, df_sampling, on='病歷號', how='inner')

# 5. 檢查合併後的 'AID' 和 'cancer' 內容
print("AID 和 cancer 的值計數:")
print(df_merged[['AID', 'cancer']].value_counts(dropna=False))
print("\n資料類型:")
print(df_merged.dtypes)

# 6. 確保 'AID' 和 'cancer' 是布林值
df_merged['AID'] = df_merged['AID'].astype(bool)
df_merged['cancer'] = df_merged['cancer'].astype(bool)

# 7. 計算 'AID' 和 'cancer' 為 True 的數量
count_AID_true = df_merged['AID'].sum()
count_cancer_true = df_merged['cancer'].sum()

# 8. 顯示結果
print(f"\n'AID' 為 True 的數量為：{count_AID_true}")
print(f"'cancer' 為 True 的數量為：{count_cancer_true}")

# 9. 過濾並顯示 'AID' 或 'cancer' 其中一個為 True 的數據（排除兩者同時為 True）
df_filtered = df_merged[df_merged['AID'] ^ df_merged['cancer']]

print("\n'AID' 或 'cancer' 其中一個為 True 的數據:")
print(df_filtered)
```
